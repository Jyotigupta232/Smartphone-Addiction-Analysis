## ⚡ High-Scale SQL Query Execution (100k+ Rows)

```python
conn = sqlite3.connect(db_path)
query = """
SELECT 
    app_category,
    COUNT(DISTINCT user_id) AS total_users,
    ROUND(AVG(session_duration_min), 2) AS avg_session_min,
    ROUND(SUM(session_duration_min), 2) AS total_duration_min
FROM app_usage_sessions_indexed
GROUP BY app_category
ORDER BY total_duration_min DESC;
"""
pd.read_sql_query(query, conn)
```
